```python
print("Sales <= 0:", (df["Sales"] <= 0).sum())

print("Quantity <= 0:", (df["Quantity"] <= 0).sum())

print(
    "Discount outside 0–1:",
    (~df["Discount"].between(0, 1)).sum()
)
```

```python
print("Total Sales:", round(df["Sales"].sum(), 2))
print("Total Profit:", round(df["Profit"].sum(), 2))
print("Total Quantity:", df["Quantity"].sum())
print("Total Orders:", df["Order ID"].nunique())
print("Unique Customers:", df["Customer ID"].nunique())
```

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")).round(2)

category_analysis
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df,x="Category",y="Sales",estimator="sum",errorbar=None)
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df,x="Category",y="Profit",estimator="sum",errorbar=None)
plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
subcategory_analysis = df.groupby("Sub-Category").agg(Total_Sales=("Sales", "sum"),Total_Profit=("Profit", "sum"),Total_Quantity=("Quantity", "sum")).round(2)

subcategory_analysis.sort_values(by="Total_Profit",ascending=False)
```

```python
profit_data = subcategory_analysis.sort_values(by="Total_Profit",ascending=True)
plt.figure(figsize=(10, 7))

sns.barplot(data=profit_data.reset_index(),x="Total_Profit",y="Sub-Category",errorbar=None)
plt.title("Profit by Sub-Category")
plt.xlabel("Total Profit")
plt.ylabel("Sub-Category")
plt.tight_layout()
plt.show()
```

```python
subcategory_analysis[subcategory_analysis["Total_Profit"] < 0].sort_values(by="Total_Profit")
```

```python
df["Month"] = df["Order Date"].dt.to_period("M")

monthly_sales = df.groupby("Month").agg(Total_Sales=("Sales", "sum"),Total_Profit=("Profit", "sum")).reset_index()
monthly_sales["Month"] = monthly_sales["Month"].astype(str)
monthly_sales.head()
```

```python
plt.figure(figsize=(14, 6))

sns.lineplot(data=monthly_sales,x="Month",y="Total_Profit",marker="o")
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print("Highest Sales Month:")
print(monthly_sales.loc[monthly_sales["Total_Sales"].idxmax()])

print("\nLowest Sales Month:")
print(monthly_sales.loc[monthly_sales["Total_Sales"].idxmin()])

print("\nHighest Profit Month:")
print(monthly_sales.loc[monthly_sales["Total_Profit"].idxmax()])

print("\nLowest Profit Month:")
print(monthly_sales.loc[monthly_sales["Total_Profit"].idxmin()])
```

```python
region_analysis = df.groupby("Region").agg(Total_Sales=("Sales", "sum"),Total_Profit=("Profit", "sum"),Total_Quantity=("Quantity", "sum")).round(2)
region_analysis = region_analysis.sort_values(
    by="Total_Sales",
    ascending=False)

region_analysis
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(data=region_analysis.reset_index(),x="Region",y="Total_Sales",errorbar=None)
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(data=region_analysis.reset_index(),x="Region",y="Total_Profit",errorbar=None)
plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
segment_analysis = df.groupby("Segment").agg(Total_Sales=("Sales", "sum"),Total_Profit=("Profit", "sum"),Total_Quantity=("Quantity", "sum"),
Unique_Customers=("Customer ID", "nunique")).round(2)

segment_analysis.sort_values(by="Total_Sales",ascending=False)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=segment_analysis.reset_index(),x="Segment",y="Total_Sales",errorbar=None)

plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=segment_analysis.reset_index(),x="Segment",y="Total_Profit",errorbar=None)
plt.title("Total Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
discount_analysis = df.groupby("Discount").agg(Total_Sales=("Sales", "sum"),Total_Profit=("Profit", "sum"),Total_Orders=("Order ID", "nunique")).round(2)

discount_analysis
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df,x="Discount",y="Profit",estimator="mean",errorbar=None)
plt.title("Average Profit by Discount Level")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.tight_layout()
plt.show()
```

```python
loss_analysis = df[df["Profit"] < 0].groupby("Discount").agg(Loss_Records=("Profit", "count"),Total_Loss=("Profit", "sum")).round(2)

loss_analysis
```

```python
customer_analysis = df.groupby(["Customer ID", "Customer Name"]).agg(Total_Sales=("Sales", "sum"),Total_Profit=("Profit", "sum"),Total_Orders=("Order ID", "nunique")
).round(2)

customer_analysis.sort_values(by="Total_Sales",ascending=False).head(10)
```

```python
customer_analysis.sort_values(by="Total_Profit",ascending=False).head(10)
```

```python
top_customers = customer_analysis.sort_values(by="Total_Profit",ascending=False).head(10).sort_values(by="Total_Profit")

plt.figure(figsize=(10, 6))
plt.barh(top_customers.index.get_level_values("Customer Name"),top_customers["Total_Profit"])
plt.title("Top 10 Most Profitable Customers")
plt.xlabel("Total Profit")
plt.ylabel("Customer Name")
plt.tight_layout()
plt.show()
```
